```python
import pandas as pd
import numpy as np
from scipy import stats# install scipy library
from scipy.stats import chi2_contingency
```

```python
Data = pd.read_csv("/content/CHI-SQUARE .csv")
Data
```

```python
# state the hypothesis
Null Hypothesis(H0): There is no Significant association b/t type of smart home device purchased and customer satisfaction level.

Alternate Hypothesis(H1): There is Significant association b/t type of smart home device purchased and customer satisfaction level.
```

```python
#to calculate the chisquare statistic value
st=pd.Series([50,80,60,30,20])
sl=pd.Series([70,100,90,50,50])
t_stat= stats.ttest_ind(st,sl) #syntax fro calculating the statistic value for 2 sample ztest.
t_statistic=t_stat.statistic  # it gives only statistical value
p_value=t_stat.pvalue         #it gives only probability value
print(p_value,t_statistic)
```

```python
from os import stat
#to determine the critical value
critical_value =stats.t.ppf(1 - 0.05, 4)
critical_value
```

```python
if t_statistic > critical_value :
    print("Accept Alternate Hypothesis(H1) and reject Null Hypothesis(H0)")
else:
    print("Accept Null Hypothesis(H0) and reject Alternate Hypothesis(H1)")
```

```python
Conclusion:  There is no Significant association b/t type of smart home device purchased and customer satisfaction level.
```
