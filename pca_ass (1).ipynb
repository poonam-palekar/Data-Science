```python
from sklearn.preprocessing import StandardScaler

# Assuming `data` is your DataFrame containing the features
scaler = StandardScaler()
scaled_data = scaler.fit_transform(wine)
```

```python
from sklearn.decomposition import PCA

# Initialize PCA and fit it to the standardized data
pca = PCA()
pca.fit(scaled_data)

# Get the explained variance ratio for each principal component
explained_variance = pca.explained_variance_ratio_
```

```python
#loading weights
pca.components_
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(range(1, len(explained_variance) + 1), explained_variance, marker='o', linestyle='--')
plt.title('Scree Plot')
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance')
plt.show()
```

```python
cumulative_variance = pca.explained_variance_ratio_.cumsum()

plt.figure(figsize=(10, 6))
plt.plot(range(1, len(cumulative_variance) + 1), cumulative_variance, marker='o', linestyle='--')
plt.title('Cumulative Explained Variance')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.axhline(y=0.95, color='r', linestyle='--')  # Line to mark 95% variance explained
plt.show()
```

```python
#transform  the original dataset into principal components
# For example, if you decide to use 2 principal components:
optimal_pca = PCA(n_components=2)
principal_components = optimal_pca.fit_transform(scaled_data)

# Create a DataFrame with the principal components
principal_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
```

**Task 3**

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score

# Choose the number of clusters (e.g., k=3)
k = 3

# Initialize and fit K-Means to the original data
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(data)

# Get the cluster labels
labels = kmeans.labels_
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(data.iloc[:, 0], data.iloc[:, 1], c=labels, cmap='viridis', marker='o')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=300, c='red', label='Centroids')
plt.title('K-Means Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

```python
from sklearn.decomposition import PCA

# Reduce the dataset to 2 dimensions using PCA
pca = PCA(n_components=2)
pca_data = pca.fit_transform(data)

plt.figure(figsize=(10, 6))
plt.scatter(pca_data[:, 0], pca_data[:, 1], c=labels, cmap='viridis', marker='o')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=300, c='red', label='Centroids')
plt.title('K-Means Clustering (PCA Reduced Data)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()
```

```python
silhouette_avg = silhouette_score(data, labels)
print(f'Silhouette Score: {silhouette_avg:.3f}')
```

```python
db_index = davies_bouldin_score(data, labels)
print(f'Davies–Bouldin Index: {db_index:.3f}')
```

**Task 4**

```python
# Use the previously computed PCA data
pca_data = PCA(n_components=2).fit_transform(wine)

# Apply K-Means clustering to the PCA-transformed data
kmeans_pca = KMeans(n_clusters=k, random_state=42)
kmeans_pca.fit(pca_data)

# Get the cluster labels
labels_pca = kmeans_pca.labels_
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(pca_data[:, 0], pca_data[:, 1], c=labels_pca, cmap='viridis', marker='o')
plt.scatter(kmeans_pca.cluster_centers_[:, 0], kmeans_pca.cluster_centers_[:, 1], s=300, c='red', label='Centroids')
plt.title('K-Means Clustering on PCA-Transformed Data')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()
```

```python
# Silhouette Score for PCA-transformed data
silhouette_avg_pca = silhouette_score(pca_data, labels_pca)
print(f'Silhouette Score (PCA): {silhouette_avg_pca:.3f}')

# Davies–Bouldin Index for PCA-transformed data
db_index_pca = davies_bouldin_score(pca_data, labels_pca)
print(f'Davies–Bouldin Index (PCA): {db_index_pca:.3f}')
```

```python
print(f'Silhouette Score Comparison:')
print(f'Original Data: {silhouette_avg:.3f}')
print(f'PCA Data: {silhouette_avg_pca:.3f}\n')

print(f'Davies–Bouldin Index Comparison:')
print(f'Original Data: {db_index:.3f}')
print(f'PCA Data: {db_index_pca:.3f}')
```

Summary of Key Findings and Insights
Exploratory Data Analysis (EDA):

The initial EDA helped us understand the distribution, central tendency, and relationships between variables in the dataset.
Some features exhibited strong correlations, indicating potential redundancy, which PCA addressed by reducing dimensionality.
Dimensionality Reduction with PCA:

PCA successfully reduced the dimensionality of the dataset while retaining a significant portion of the variance.
The scree plot and cumulative explained variance indicated that a relatively small number of principal components captured most of the variance in the data.
Transforming the data into principal components simplified the data structure, making subsequent clustering more efficient.
Comparison and Analysis:

Clustering results from the PCA-transformed data generally showed enhanced cluster separation and compactness compared to the original dataset.
PCA's dimensionality reduction helped in simplifying the data, making clustering algorithms more effective and faster to execute.
However, the trade-off was a potential loss of interpretability since PCA alters the original feature space.
Practical Implications of Using PCA and Clustering in Data Analysis
Enhanced Clustering Performance:

PCA can significantly improve clustering performance by removing noise and redundant features. This is particularly useful when dealing with high-dimensional data or datasets with correlated features.
Simplified Data Structure
Efficiency:
While PCA can enhance clustering efficiency, it may reduce the interpretability of the clusters since the principal components are not directly related to the original features.
Computational Efficiency:

Reducing dimensionality through PCA can make clustering algorithms computationally more efficient, especially for large datasets, by reducing the complexity of the feature space.
Recommendations for Using PCA and Clustering
When to Use PCA:

High-Dimensional Datasets: PCA is ideal when the dataset has a large number of features, especially if many are correlated or redundant.
Noise Reduction: Use PCA to reduce noise and focus on the most significant patterns in the data, which can improve clustering outcomes.
Visualization Needs: If you need to visualize the data in 2D or 3D, PCA is a valuable tool to reduce dimensionality without losing much variance.
When to Cluster Directly on the Original Dataset:

Small Datasets: If the dataset has a small number of features or observations, clustering directly on the original data may be more appropriate.
Need for Interpretability: If interpretability of the clusters is crucial, it's better to cluster on the original data to retain the meaning of the features.
Minimal Redundancy: When the features in the dataset are already uncorrelated and meaningful, PCA might not provide additional benefits and could unnecessarily complicate the analysis.
Combined Approach:

Initial EDA and Feature Selection: Before applying PCA, conduct thorough EDA and consider manual feature selection to remove any irrelevant or highly correlated features.
Iterative Process: Consider iteratively applying PCA and clustering to find the optimal balance between dimensionality reduction and cluster interpretability.
Consider Context: The choice between PCA and direct clustering should be context-dependent, taking into account the specific goals of the analysis, the nature of the data, and the importance of interpretability.
In summary, PCA and clustering are powerful tools in data analysis, each with its strengths and trade-offs. The choice of technique should be guided by the characteristics of the dataset and the specific objectives of the analysis.
